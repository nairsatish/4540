```python
import numpy as np
import pandas as pd
%pip install scikit-learn
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Step 1: Create the dataset
data = np.array([[1, 2], [2, 4], [3, 6], [7, 8]])
df = pd.DataFrame(data, columns=['x', 'y'])

# Step 2: Standardize the data
scaler = StandardScaler()
standardized_data = scaler.fit_transform(df)


# Step 3: Calculate the covariance matrix
cov_matrix = np.cov(standardized_data.T)

# Step 4: Calculate the eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

# Step 5: Sort eigenvalues and eigenvectors
sorted_index = np.argsort(eigenvalues)[::-1]
sorted_eigenvalues = eigenvalues[sorted_index]
sorted_eigenvectors = eigenvectors[:, sorted_index]

# Step 6: Transform the data
transformed_data = np.dot(standardized_data, sorted_eigenvectors)

# Print results
print("Original Data:\n", df)
print("\nStandardized Data:\n", standardized_data)
print("\nCovariance Matrix:\n", cov_matrix)
print("\nEigenvalues:\n", sorted_eigenvalues)
print("\nEigenvectors:\n", sorted_eigenvectors)
print("\nTransformed Data:\n", transformed_data)
```
