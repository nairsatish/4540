```python
import numpy as np
import pandas as pd
%pip install scikit-learn
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

def center_matrix(matrix):
    # Convert the input to a NumPy array
    matrix = np.array(matrix)
    
    # Calculate the mean of each column
    mean_x = np.mean(matrix[:, 0])
    mean_y = np.mean(matrix[:, 1])
    
    # Subtract the mean from each column
    centered_matrix = matrix - np.array([mean_x, mean_y])
    
    return centered_matrix

#Step 2B: You can also Standardize the data
def standardize_matrix(matrix):
    # Convert the input to a NumPy array
    matrix = np.array(matrix)
    
    # Calculate the mean and standard deviation of each column
    mean = np.mean(matrix, axis=0)
    std_dev = np.std(matrix, axis=0)
    
    # Standardize the data
    standardized_data = (matrix - mean) / std_dev
    
    return standardized_matrix


# Step 1: Create the dataset
matrix = [
    [1,2],
    [2,4],
    [3,6],
    [7,8]
]

# Step 2: Center the data
centered_matrix = center_matrix(matrix)


# Another approach Step 2: Standardize the data
#scaler = StandardScaler()
#standardized_data = scaler.fit_transform(df)

# Step 3: Calculate the covariance matrix
cov_matrix = np.cov(centered_matrix.T)

# Step 4: Calculate the eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

# Step 5: Sort eigenvalues and eigenvectors
sorted_index = np.argsort(eigenvalues)[::-1]
sorted_eigenvalues = eigenvalues[sorted_index]
sorted_eigenvectors = eigenvectors[:, sorted_index]

# Step 6: Transform the data
transformed_data = np.dot(standardized_data, sorted_eigenvectors)

# Print results
print("Original Data:\n", np.array(matrix))
print("Centered Matrix:\n", centered_matrix)
print("Covariance Matrix:\n", cov_matrix)
print("Eigenvalues:\n", sorted_eigenvalues)
print("Eigenvectors:\n", sorted_eigenvectors)
print("Transformed Data:\n", transformed_data)
```
